```python
import cv2
import os
```

```python
class StereoVisionDistance:
    def __init__(self, baseline=5.0):
        self.baseline = baseline
        self.focal = None

    def calc_pmets(self, left_img_path, left_x_cnorm, right_x_cnorm):
        img = cv2.imread(left_img_path)

        if img is None:
            raise ValueError(f"Image at path {left_img_path} could not be loaded.")

        img_h, img_w = img.shape
        x_lp = left_x_cnorm * img_w
        x_rp = right_x_cnorm * img_w

        disparity = x_lp - x_rp
        return img_w, round(disparity, 2)

    def calibrate_sc(self, pdisparity, known_distance):
        self.focal = (pdisparity * known_distance) / self.baseline
        return round(self.focal, 2)

    def distance_estimation(self, pdisparity):
        if self.focal is None:
            raise ValueError("Focal length not calibrated.")

        if pdisparity <= 0: 
            raise ValueError("Pixel disparity must be greater than zero.")

        distance = (self.focal * self.baseline) / pdisparity
        return round(distance, 2)
```
